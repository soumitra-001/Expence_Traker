```python
price = 50
s5 = f"rice cost is {price}"
s6 = f"rice cost is {price:.2f}" #{} are used as placeholder for variable
print(s5)
print(s6)
s7 = f"2 kg rice cost is {price* 2:.2f}"
print(s7)
```

```python
score = 94.56223
s10 = f"score :{score:>8.2f} | status : {'PASS':^10}"
# after decimal 2 digit , 8 is minimum 8 char,^ is center align
print(s10)
```

```python
name = 'AI'
s11 = f"{name:*^10}"
print(s11)
# outcome is 10 character length
# * fill character with *
# ^ is for center alignment
```

```python
name = 'AI'
s11 = f"{name:+>8}"
print(s11)
# outcome is 8 character length
# + fill character with *
# > is for right alignment
```

```python
name = 'AI'
s11 = f"{name:-<11}"
print(s11)
# outcome is 11 character length
# - fill character with *
# < is for left alignment
```

```python
x = -45.68
s13 = f"result:{x:=10.2f}"
print(s13)
```

```python
x = -45.68
s13 = f"result:{x:*=10.2f}" # = is for sign-aware padding
print(s13)
type(s13)
```
